```python
%sql
-- Who am I, as far as the functions can tell?
SELECT current_user()                          AS me,
       is_account_group_member('data_engineers') AS in_engineers,
       is_account_group_member('compliance')     AS in_compliance,
       is_account_group_member('analysts')       AS in_analysts;
```

```python
%sql
--- I am currently part of the data engineer and compliance group, so I should see the full SSN

-- Column mask: compliance members see full SSNs, others see ***-**-1234.
SELECT customer_id, ssn, state
FROM dev_kyc.gold.dim_customer
WHERE is_current
LIMIT 10;
```

```python
%sql
-- Row filter: engineers and compliance see all states; analysts only FL and GA.
SELECT customer_state, COUNT(*) AS rows
FROM dev_kyc.gold.customer_risk_daily
GROUP BY customer_state
ORDER BY customer_state;
```
